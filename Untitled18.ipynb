```python
import os
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from flask import Flask, request, jsonify, render_template
from flask_cors import CORS
from pyngrok import ngrok
import threading
import time

#  archivo HTML
os.makedirs("templates", exist_ok=True)
with open("templates/index.html", "w") as f:
    f.write("""
<!DOCTYPE html>
<html lang="es">
<head>
  <meta charset="UTF-8">
  <title>Libro Vibes</title>
  <style>
    body {
      font-family: 'Segoe UI', sans-serif;
      background-color: #f0f4c3;
      margin: 0;
      padding: 0;
    }
    .container {
      max-width: 600px;
      margin: 50px auto;
      background: white;
      border-radius: 15px;
      box-shadow: 0 0 20px rgba(126,87,194,0.2);
      padding: 30px;
      border: 2px solid #7e57c2;
    }
    h1 {
      text-align: center;
      color: #7e57c2;
      font-family: 'Courier New', cursive;
    }
    input[type="text"] {
      width: 80%;
      padding: 12px;
      margin-right: 10px;
      border-radius: 20px;
      border: 2px dashed #7e57c2;
      font-size: 16px;
    }
    button {
      padding: 12px 20px;
      border: none;
      background-color: #7e57c2;
      color: white;
      border-radius: 20px;
      cursor: pointer;
      font-size: 16px;
    }
    #resultado {
      margin-top: 20px;
    }
    .libro {
      background-color: rgba(126, 87, 194, 0.05);
      border-left: 5px solid #7e57c2;
      margin-bottom: 15px;
      padding: 10px 15px;
      border-radius: 0 10px 10px 0;
    }
    .libro p {
      margin: 5px 0;
    }
    .error {
      color: red;
      font-weight: bold;
    }
  </style>
</head>
<body>
  <div class="container">
    <h1>Libro Vibes 📚</h1>
    <div style="text-align:center;">
      <input type="text" id="inputLibro" placeholder="Escribe el nombre de un libro">
      <button onclick="buscar()">Buscar</button>
    </div>
    <div id="resultado"></div>
  </div>

  <script>
    function buscar() {
      const libro = document.getElementById("inputLibro").value.trim();
      const resDiv = document.getElementById("resultado");
      resDiv.innerHTML = "";
      if (!libro) {
        resDiv.innerHTML = "<p class='error'>⚠️ Por favor ingresa un título</p>";
        return;
      }
      fetch("/recommend", {
        method: "POST",
        headers: { "Content-Type": "application/json" },
        body: JSON.stringify({ libro: libro })
      })
      .then(response => response.json())
      .then(data => {
        if (data.error) {
          resDiv.innerHTML = "<p class='error'>❌ " + data.error + "</p>";
        } else {
          let html = "";
          data.forEach((item, i) => {
            html += `<div class="libro">
                      <p><strong>${i + 1}. ${item.Book}</strong></p>
                      <p>Autor: ${item.Author || 'Desconocido'}</p>
                      <p>⭐ Calificación: ${item.Avg_Rating}</p>
                    </div>`;
          });
          resDiv.innerHTML = html;
        }
      })
      .catch(err => {
        resDiv.innerHTML = "<p class='error'>Error de red: " + err + "</p>";
      });
    }
  </script>
</body>
</html>
""")


# Datos
df = pd.read_excel("/content/goodreads_data (1).xlsx", engine="openpyxl")
df['Genres_list'] = df['Genres'].str.split(',\s*').fillna('').apply(lambda x: [g.strip().capitalize() for g in x])
df['Contenido'] = df['Book'].astype(str) + ' ' + df['Genres_list'].apply(' '.join) + ' ' + df['Description'].astype(str)

vectorizer = TfidfVectorizer(stop_words='english', max_features=5000)
X = vectorizer.fit_transform(df['Contenido'])
similitud = cosine_similarity(X)

# Flask app
app = Flask(__name__)
CORS(app)

@app.route('/')
def index():
    return render_template("index.html")

@app.route('/recommend', methods=['POST'])
def recommend():
    data = request.json
    libro = data.get('libro', '').strip().lower()
    mask = df['Book'].str.strip().str.lower() == libro

    if not mask.any():
        return jsonify({"error": "Libro no encontrado"}), 404

    idx = df[mask].index[0]
    scores = sorted(enumerate(similitud[idx]), key=lambda x: x[1], reverse=True)[1:6]
    resultados = df.iloc[[i[0] for i in scores]][['Book', 'Author', 'Avg_Rating']]
    return jsonify(resultados.to_dict('records'))

# Iniciar Flask en hilo
def run():
    app.run(port=5002, use_reloader=False)

threading.Thread(target=run).start()

# Tiempo para que Flask arranque
time.sleep(2)

# Conectar ngrok DESPUÉS
ngrok.set_auth_token("2w3OkjAIRDq3JjKJpnZy9FDrToK_6QgiYmPuyUgeuzmjmFq3n")
public_url = ngrok.connect(5002)
print("Tu app está disponible en:", public_url)
```
